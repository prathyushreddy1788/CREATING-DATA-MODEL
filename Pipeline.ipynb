```sql
%sql
CREATE OR REFRESH STREAMING TABLE bronze_sales AS
SELECT * FROM STREAM(my_catalog.my_db1.sales_fact);

CREATE OR REFRESH STREAMING TABLE silver_sales AS
SELECT
  sales_id,
  date_id,
  product_id,
  customer_id,
  quantity,
  total_amount
FROM STREAM(bronze_sales)
WHERE quantity > 0 AND total_amount > 0;

CREATE OR REFRESH MATERIALIZED VIEW gold_sales_by_product AS
SELECT
  product_id,
  SUM(quantity) AS total_quantity,
  SUM(total_amount) AS total_sales
FROM silver_sales
GROUP BY product_id;
```
