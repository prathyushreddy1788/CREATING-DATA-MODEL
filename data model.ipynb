```python
%sql

CREATE TABLE my_catalog.my_db1.date_dim (
  date_id INT PRIMARY KEY,
  date DATE,
  year INT,
  month INT,
  day INT
) USING DELTA;

CREATE TABLE my_catalog.my_db1.product_dim (
  product_id INT PRIMARY KEY,
  product_name STRING,
  category STRING,
  price DECIMAL(10,2)
) USING DELTA;

CREATE TABLE my_catalog.my_db1.customer_dim (
  customer_id INT PRIMARY KEY,
  customer_name STRING,
  region STRING
) USING DELTA;
```

```python
%sql
CREATE OR REPLACE TABLE my_catalog.my_db1.sales_fact (
  sales_id INT PRIMARY KEY,
  date_id INT REFERENCES my_catalog.my_db1.date_dim(date_id),
  product_id INT REFERENCES my_catalog.my_db1.product_dim(product_id),
  customer_id INT REFERENCES my_catalog.my_db1.customer_dim(customer_id),
  quantity INT,
  total_amount DECIMAL(10,2)
) USING DELTA;
```
